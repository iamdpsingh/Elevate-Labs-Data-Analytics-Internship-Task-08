```python
import pandas as pd

df = pd.read_csv('train.csv')

df.head(), df.info()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')
```

```python
missing_before = df.isnull().sum()
```

```python

df_cleaned = df.dropna(subset=['Order Date', 'Sales'])
```

```python
df_cleaned['Postal Code'] = df_cleaned['Postal Code'].fillna(-1)
```

```python
df_cleaned['OrderMonthYear'] = df_cleaned['Order Date'].dt.strftime('%b-%Y')
```

```python
missing_after = df_cleaned.isnull().sum()
```
